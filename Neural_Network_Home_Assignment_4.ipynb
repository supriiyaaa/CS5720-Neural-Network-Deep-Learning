```python


import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Download required NLTK resources
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

def preprocess_text(sentence):
    # Step 1: Tokenize the sentence
    tokens = word_tokenize(sentence)
    print("Original Tokens:", tokens)

    # Step 2: Remove stopwords
    stop_words = set(stopwords.words("english"))

    filtered_tokens = [
        word for word in tokens
        if word.lower() not in stop_words
    ]
    print("Tokens Without Stopwords:", filtered_tokens)

    # Step 3: Apply Porter stemming
    stemmer = PorterStemmer()

    stemmed_words = [
        stemmer.stem(word) for word in filtered_tokens
    ]
    print("Stemmed Words:", stemmed_words)

    return tokens, filtered_tokens, stemmed_words


sentence = "NLP techniques are used in virtual assistants like Alexa and Siri."

preprocess_text(sentence)
```

**1. What is the difference between stemming and lemmatization?**

Stemming removes word endings to obtain a root form, which may not be a valid word. For example, "running" becomes "run". Lemmatization uses vocabulary and linguistic analysis to convert words into their dictionary form. For example, "running" becomes "run" when used as a verb.

**2. Why is removing stopwords useful, and when can it be harmful?**

Removing stopwords reduces unnecessary words and improves efficiency in tasks like text classification. However, it can be harmful in sentiment analysis because removing words like "not" may change the meaning of a sentence.


```python


import spacy

# Load pretrained English NLP model
nlp = spacy.load("en_core_web_sm")

def extract_entities(sentence):
    # Process the sentence using spaCy
    doc = nlp(sentence)

    # Extract and display named entities
    for ent in doc.ents:
        print(
            f"Entity: {ent.text} | "
            f"Label: {ent.label_} | "
            f"Start: {ent.start_char} | "
            f"End: {ent.end_char}"
        )

sentence = (
    " LeBron James as the NBA player of the "
    "United States in Laker and won the NBA "
    "Final MVP in 2020."
)

extract_entities(sentence)
```

**1. How does NER differ from POS tagging in NLP?**

Named Entity Recognition (NER) identifies entities such as people, organizations, locations, and dates. Part-of-Speech (POS) tagging identifies grammatical roles such as nouns, verbs, and adjectives.

```python

import numpy as np

def softmax(x):
    # Apply softmax along each row
    exp_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=-1, keepdims=True)


def scaled_dot_product_attention(Q, K, V):
    # Step 1: Compute dot product of Q and K transpose
    scores = np.matmul(Q, K.T)

    # Step 2: Scale scores by square root of key dimension
    d_k = K.shape[-1]
    scaled_scores = scores / np.sqrt(d_k)

    # Step 3: Apply softmax to obtain attention weights
    attention_weights = softmax(scaled_scores)

    # Step 4: Multiply attention weights by V
    output = np.matmul(attention_weights, V)

    return attention_weights, output


# Given input matrices
Q = np.array([[1, 0, 1, 0],
              [0, 1, 0, 1]])

K = np.array([[1, 0, 1, 0],
              [0, 1, 0, 1]])

V = np.array([[1, 2, 3, 4],
              [5, 6, 7, 8]])

# Calculate attention
weights, output = scaled_dot_product_attention(Q, K, V)

# Display results
print("Attention Weights:")
print(np.round(weights, 3))

print("\nFinal Output Matrix:")
print(np.round(output, 3))
```

**1. Why do we divide the attention score by √d in scaled dot-product attention?**

Dividing by √d prevents attention scores from becoming too large when the key dimension increases. This keeps the softmax function stable and helps prevent very small gradients during training.

**2. How does self-attention help the model understand relationships between words in a sentence?**

Self-attention allows each word to focus on other relevant words in a sentence, regardless of their positions. This helps the model understand context, meaning, and long-distance relationships between words.
